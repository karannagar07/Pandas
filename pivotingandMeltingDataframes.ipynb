```python
import pandas as pd
dict = {"keys" : ["k1", "k2", "k3"],
         "names":["john", "ben", "david"],
         "houses":["red", "blue", "green"],
         "grades":["3rd", "8th", "9th"]}
df = pd.DataFrame(dict)
print(df)
print(df.pivot(index="keys", columns="names", values="houses"))#rows converted into columns
```

```python
import pandas as pd
dict = {
         "names":["john", "ben", "david"],
         "houses":["red", "blue", "green"],
         "grades":["3rd", "8th", "9th"]}
df = pd.DataFrame(dict)
print(df)
print(pd.melt(df,id_vars=["names"],value_vars=["houses","grades"]))
```
