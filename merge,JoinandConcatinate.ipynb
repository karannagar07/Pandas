```python
# First DataFrame

import pandas as pd

# First DataFrame
df1 = pd.DataFrame({
    'emp_id': [101, 102, 103],
    'name': ['Karan', 'Diya', 'Aman'],
    'age': [23, 22, 25]
})
print(df1)

# Second DataFrame
df2 = pd.DataFrame({
    'emp_id': [101, 106, 103],
    'salary': [30000, 28000, 35000]
})
print(df2)

# Merge both DataFrames on emp_id
final_df = pd.merge(df1, df2, on='emp_id')

print(final_df)
```

```python
# how me jo left h iska mtlb h left table ke sab aayenge 
#how me use kr skte h ->left,right,inner,outer
final_df = pd.merge(df1, df2, on='emp_id',how="left")

print(final_df)
```

```python
import pandas as pd

# First DataFrame

df1 = pd.DataFrame({
    'emp_id': [101, 102, 103],
    'name': ['Karan', 'Diya', 'Aman'],
    'age': [23, 22, 25]
})
df2 = pd.DataFrame({
    'emp_id': [104, 105, 106],
    'name': ['varun', 'jiya', 'ram'],
    'age': [23, 22, 25]
})
final_df = pd.concat([df1, df2])# agar yha axis=1 hoga to concat row k basis p hoga
print(final_df)
```
